```python
# checking  possible values in categorical columns
for column in ["study_mode", "python_confidence", "group_code"]:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python
# Checking  the date column, it has different date formats
df["last_activity_parsed"] = pd.to_datetime(
    df["last_activity"],
    errors="coerce",
    dayfirst=True
)

df[["last_activity", "last_activity_parsed"]].head(10)
```

```python
# Count dates that could not be read
df["last_activity_parsed"].isna().sum()
```
